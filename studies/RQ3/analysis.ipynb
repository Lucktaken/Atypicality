# RQ3: 地理/发展水平调节效应

研究问题：地理因素（半球/发展水平）如何调节数据使用非典型性与学术成功的关系？

因变量：Atypicality
核心自变量：H_index 等
调节变量：Hemisphere, Development_Status

模型：分组 OLS + NB + 鲁棒性

```python
import sys, os
sys.path.insert(0, os.path.join(os.getcwd(), '..', '..'))

from src.data_loader import load_config
from src.regression import run_ols, run_nb
from src.visualization import plot_butterfly_chart, save_fig
import pandas as pd
```

```python
config = load_config()
df = pd.read_csv(os.path.join('studies', 'RQ3', 'rq3_regression_ready.csv'))
print(f'数据维度: {df.shape}')
df.head()
```

## 1. 分组 OLS 回归

```python
x_var = 'H_index'
controls = ['Academic_Age', 'Team_Size', 'Topic_Diversity']
controls = [c for c in controls if c in df.columns]
y_var = 'Atypicality'

group_results = {}
for group_col in ['Hemisphere', 'Development_Status']:
    if group_col in df.columns:
        for group_val in df[group_col].dropna().unique():
            subset = df[df[group_col] == group_val]
            label = f'{group_col}={group_val}'
            group_results[label] = run_ols(subset, y_var, x_var, controls)
            if group_results[label]['converged']:
                print(f'{label}: coef={group_results[label]["coef"]:.4f}, p={group_results[label]["pvalue"]:.4f}')
```

## 2. Butterfly Chart
