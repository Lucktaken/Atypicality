# RQ6: 数据使用模式的 Embedding 分析

研究问题：学者使用数据的模式如何？不同职业阶段的学者在数据使用上有何差异？

方法：Word2Vec + 聚类 + 描述性统计

```python
import sys, os
sys.path.insert(0, os.path.join(os.getcwd(), '..', '..'))

from src.data_loader import load_config
import pandas as pd
import numpy as np
```

```python
config = load_config()
df = pd.read_csv(os.path.join('studies', 'RQ6', 'rq6_analysis_ready.csv'))
print(f'数据维度: {df.shape}')
df.head()
```

## 1. 加载 Word2Vec 模型

```python
from gensim.models import Word2Vec

model_path = os.path.join(config['paths']['interim_q2'], 'dataset_word2vec.model')
if os.path.exists(model_path):
    w2v_model = Word2Vec.load(model_path)
    print(f'词表大小: {len(w2v_model.wv)}')
else:
    print('Word2Vec 模型未找到，请先运行训练')
```

## 2. 描述性统计：不同职业阶段的数据使用

```python
if 'Academic_Age' in df.columns:
    df['Career_Stage'] = pd.cut(df['Academic_Age'], bins=[0, 5, 15, 100],
                                labels=['Early', 'Mid', 'Senior'])
    stage_stats = df.groupby('Career_Stage').agg(
        N=('AuthorID', 'count'),
        Mean_Atypicality=('Atypicality_of_datasets_original_1', 'mean'),
        Median_Atypicality=('Atypicality_of_datasets_original_1', 'median'),
    )
    print(stage_stats)
```
