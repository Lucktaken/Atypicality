# RQ4: 数据使用非典型性的决定因素

研究问题：什么因素决定了数据使用的非典型性？

因变量：Atypicality_of_datasets_original_1
自变量：H_index, Academic_Age, Topic_Diversity, Average_Team_Size 等

模型：OLS

```python
import sys, os
sys.path.insert(0, os.path.join(os.getcwd(), '..', '..'))

from src.data_loader import load_config
from src.regression import run_ols, save_results
from src.visualization import plot_forest_chart, save_fig
import pandas as pd
```

```python
config = load_config()
df = pd.read_csv(os.path.join('studies', 'RQ4', 'rq4_regression_ready.csv'))
print(f'数据维度: {df.shape}')
df.head()
```

## 1. OLS 回归：各因素对 Atypicality 的影响

```python
y_var = 'Atypicality_of_datasets_original_1'
x_candidates = ['H_index', 'Academic_Age', 'Topic_Diversity', 'Average_Team_Size',
               'Avg_Citation_Without_Self', 'unique_data_count', 'repeated_data_usage']
x_candidates = [x for x in x_candidates if x in df.columns]

results = {}
for x_var in x_candidates:
    other_controls = [c for c in x_candidates if c != x_var]
    results[x_var] = run_ols(df, y_var, x_var, other_controls)
    if results[x_var]['converged']:
        print(f'{x_var}: coef={results[x_var]["coef"]:.4f}, p={results[x_var]["pvalue"]:.4f}')
```

## 2. Forest Plot

```python
save_results(results, os.path.join('studies', 'RQ4', 'results', 'rq4_regression_results.csv'))
fig = plot_forest_chart(results, title='RQ4: Determinants of Data Atypicality')
save_fig(fig, os.path.join('studies', 'RQ4', 'figures', 'rq4_forest_plot.pdf'))
plt.show()
```
